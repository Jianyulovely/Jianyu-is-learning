- web网站由三个核心部分：前端程序、服务端程序、数据库
- 前端程序：负责前端页面的展示，网页由HTML（网页结构）、CSS（网页的表现）、JavaScript（网页的动作）三个部分组成
    HTML负责所有内容的陈列，CSS负责进行内容的布局排版，优化美观，Javascript负责动态交互效果
- 服务端程序：可以基于Python中的Django、Flask或 FastAPI 进行开发

API接口：应用程序编程接口(Application Programming Interface)，就是对外提供的功能入口，供别人来调用  
```uvicorn.run("api:app", host="0.0.0.0", port=8000, reload=True)```  
host为0.0.0.0意味着在任意主机上都可以访问这个服务端，访问的端口号是8000

uvicorn是一个python中轻量级服务器
