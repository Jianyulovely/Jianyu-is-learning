# 函数基础

注意：函数可以没有返回值 必须先定义后执行  

````python
print("helloworld")

def out_line():
    print('----------------------')

out_line()
````

### 调用函数时必须加上括号，否则就是将函数本身作为一个对象进行传输
这个例子中 **a** 是函数所在的地址，说明 out_line 就是一个函数对象，相当于一个变量，它里面存着函数  
由于函数本身没有返回值，所以 **b** 是None

````python
a = out_line
print(a)

b = out_line()
print(b)
````

### **return** 只有返回功能，没有输出打印功能
- 形参（形式参数）：函数定义时括号里的参数（如半径r），只能在函数内使用（局部变量）
- 实参（实际参数）：函数在实际调用时传入的参数（如2.5）

````python
def circle_area(r: float) -> float:
    """
    根据圆的半径计算面积
    :parm r: 圆的半径
    :return: 圆的面积
    """
    area = r**2 * 3.14
    return area

s = circle_area(2.5)
print(s)
````

多个返回值结果用逗号隔开，在调用之后被封装到**元组**当中  
可以对元组进行**解包**，用两个变量接收

````python
def circle_circumference(r: float) -> float:
    """根据圆的半径计算周长

    Args:
        r (float): 根据圆的半径

    Returns:
        float: 圆的周长
    """
    cir = 3.14 * 2 * r
    return cir

def circle_a_c(r: float) -> tuple[float, float]:
    """根据圆的半径计算面积和周长

    Args:
        r (float): 圆的半径

    Returns:
        tuple[float, float]: 面积, 周长
    """
    area = circle_area(r)
    c = circle_circumference(r)
    return round(area, 1), round(c, 1)
````

````python
circle_info = circle_a_c(1.8)
print(circle_info)

area, circum = circle_a_c(1.8)
print(area, circum)
````

### 函数嵌套调用
- 调用遵循栈结构，最后调用的函数最先返回**LIFO(Last In First Out, 后进先出)**  
程序总是先执行栈顶的代码：  


````python
def function_a():
    print("a ... before")
    function_b()
    print("a ... after")

def function_b():
    print("b ... before")
    function_c()
    print("b ... after")

def function_c():
    print("c ... ")
````

````python
def trangle_area(s: float, h: float) -> float:
    """根据底和高计算三角形面积

    Args:
        s (float): 三角形的底边长度
        h (float): 三角形的高

    Returns:
        float: 三角形面积
    """
    area = s * h / 2
    return round(area, 1)

print(trangle_area(2, 3))
````

````python
def num_yuanyin(s: str) -> int:
    """计算传入字符串中元音个数

    Args:
        s (str): 待判定字符串

    Returns:
        int: 元音数量
    """
    num = 0
    yuanyin = "aeiouAEIOU"
    for c in s:
        if c in yuanyin:
            num += 1
    return num

print(num_yuanyin("aoisnErow"))
````

````python
def get_grade_info(grade_list: list[float]) -> tuple[float, float, float]:
    """根据班级成绩列表计算最高分、最低分、平均分（1位小数）

    Args:
        grade_list (list[float]): 成绩列表

    Returns:
        tuple[float, float, float]: _description_
    """
    max_score = max(grade_list)
    min_score = min(grade_list)
    avg_score = sum(grade_list) / len(grade_list)
    return max_score, min_score, round(avg_score, 1)
````
