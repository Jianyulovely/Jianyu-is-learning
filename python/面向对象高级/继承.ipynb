### 继承
- 描述两个类之间的关系，子类继承父类，就能够获取到父类的属性和方法  
    所有类都有一个父类，根类是object类
- 重新：指子类继承父类后，可以在子类中重新定义父类中已有的方法（方法名相同），  
从而用子类的实现来覆盖父类的实现

````python
class Car:
    def __init__(self, c_brand, c_name, c_year, c_owner):
        self.brand = c_brand
        self.name = c_name
        self.year = c_year
        # 私有属性
        self.__owner = c_owner
    
    def start(self):
        print(f"{self.__owner} 的 {self.brand} {self.name} {self.year} 的车启动了")
        
        # 私有方法
        self.__control_fuel()
        
    def stop(self):
        print(f"{self.brand} {self.name} {self.year} 的车停止了")

    def __control_fuel(self):
        print(f"{self.brand} {self.name} {self.year} 的车控制了燃油")
    
    def get_owner(self):
        return self.__owner[0:1] + "*"
    
    def charge(self):
        print(f"{self.brand} {self.name} {self.year} 的车正在加油")
````

注意：如果子类在重写父类的方法时，需要调用父类的方法，  
可以通过```父类名.方法名(self) / super().方法名()```的方式进行调用。 例子如下：
```
class User:
    def login(self):
        print("验证用户名和密码")


class Admin(User):
    def login(self):
        super().login()    # 先完成父类的登录验证
        print("加载管理员权限")
```

````python
class FuelCar(Car):
    pass

class ElectricCar(Car):
    def charge(self):
        print(f"{self.brand} {self.name} {self.year} 的车正在充电")
````

````python
fuel_car1 = FuelCar("BMW", "宝马", 2023, "张三")
electric_car1 = ElectricCar("Xiaomi", "Su7", 2025, "李四")

fuel_car1.start()
fuel_car1.stop()
print(fuel_car1.get_owner())
fuel_car1.charge()
````

````python
electric_car1.charge()
````

### 多继承
- 一个子类同时继承了多个父类
- MRO：Method Resolution Order，方法解析顺序  
当一个类继承多个父类时，默认优先使用第一个父类中的同名属性或方法
可以通过 ```类名.__mro__```或者 ```类名.mro()``` 来查看MRO
```

````python
class HuaweiAiDriving:
    def __init__(self, version="1.0"):
        self.version = version
    
    def run(self):
        print(f"华为AI驾驶系统 {self.version} 正在运行")


# 问界汽车
class WenJieCar(Car, HuaweiAiDriving):
    def __init__(self, brand, name, year, owner, version="1.0"):
        super().__init__(brand, name, year, owner)
        HuaweiAiDriving.__init__(self, version)
    
    def run(self):
        Car.start(self)
        HuaweiAiDriving.run(self)
````

````python
c = WenJieCar("奔驰", "奔驰S级", 2023, "张三")
c.run()
````
