### 多态
- 指同一个方法，具有不同的形态、行为和表现

```python
class Car:
    def __init__(self, c_brand, c_name, c_year):
        self.brand = c_brand
        self.name = c_name
        self.year = c_year
    
    def start(self):
        print(f"{self.__owner} 的 {self.brand} {self.name} {self.year} 的车启动了")
        
    def stop(self):
        print(f"{self.brand} {self.name} {self.year} 的车停止了")
    
    def charge(self):
        print(f"{self.brand} {self.name} {self.year} 的车正在加油")


class FuelCar(Car):
    def __init__(self, c_brand, c_name, c_year):
        super().__init__(c_brand, c_name, c_year)
        self.fuel = "燃油"

class ElectricCar(Car):
    def __init__(self, c_brand, c_name, c_year):
        super().__init__(c_brand, c_name, c_year)
    
    def charge(self):
        print(f"{self.brand} {self.name} {self.year} 的车正在充电")
       
```

函数的参数类型是父类型，而实际调用的时候输入的是子类型。所以不同的子类型会有不同的行为。

```python
def handle_charge(car: Car):
    car.charge()

c1 = ElectricCar("特斯拉", "Model S", 2023)
handle_charge(c1)
```

### 多态-鸭子类型
- 鸭子类型是指，一个对象如果像鸭子，走起像鸭子，叫起来像鸭子，那么它就是鸭子。
- 在鸭子类型中，我们关注的是对象的行为，而不是对象的类型。
- 不依赖于继承关系，只要对象有响应的方法就能使用。就像例子中的三种动物都有swimming的方法
