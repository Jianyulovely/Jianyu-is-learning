### 封装
- 封装就是把数据（属性）和操作数据的函数（方法）捆绑在一起，形成一个独立的单元（类），并隐藏内部的实现细节，只对外暴露必要的功能（方法）  
重点是一个类可以有**公有属性和方法**，也可以有**私有属性和方法**  
- 私有的属性和方法只能在类的内部使用，python中没有真正的私有机制  
- 注意事项：这种私有机制是一种约定，外部可以通过特殊的方式访问到私有属性和方法，但是不建议这样做  
比如：
```print(car1._Car__owner)```  
```print(car1._Car__control_fuel())```


````python
class Car:
    def __init__(self, c_brand, c_name, c_year, c_owner):
        self.brand = c_brand
        self.name = c_name
        self.year = c_year
        # 私有属性
        self.__owner = c_owner
    
    def start(self):
        print(f"{self.__owner} 的 {self.brand} {self.name} {self.year} 的车启动了")
        
        # 私有方法
        self.__control_fuel()
        
    def stop(self):
        print(f"{self.brand} {self.name} {self.year} 的车停止了")

    def __control_fuel(self):
        print(f"{self.brand} {self.name} {self.year} 的车控制了燃油")
    
    def get_owner(self):
        return self.__owner[0:1] + "*"
````

````python
car1 = Car("奔驰", "奔驰S级", 2023, "张三")
car1.start()
car1.stop()
````

````python
car1.__control_fuel()
````

````python
print(car1.__owner)
````

````python
print(car1.get_owner())
````
