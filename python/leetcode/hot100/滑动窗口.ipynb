3. 给定一个字符串 s ，请你找出其中不含有重复字符的 最长 子串 的长度。  
示例 1:  
输入: s = "abcabcbb"  
输出: 3   
解释: 因为无重复字符的最长子串是 "abc"，所以其长度为 3。注意 "bca" 和 "cab" 也是正确答案。  
示例 2:  
输入: s = "bbbbb"  
输出: 1  
解释: 因为无重复字符的最长子串是 "b"，所以其长度为 1。  
示例 3:  
输入: s = "pwwkew"  
输出: 3  
解释: 因为无重复字符的最长子串是 "wke"，所以其长度为 3。  
请注意，你的答案必须是 子串 的长度，"pwke" 是一个子序列，不是子串。
 

提示：

0 <= s.length <= 5 * 104
s 由英文字母、数字、符号和空格组成

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        tmp = []
        for e in s:
            tmp.append(e)
        if len(tmp) == 0:
            return 0
        if len(tmp) == 1:
            return 1
        max_list = []
        for i in range(len(s)-1):
            max_len = 1
            tmp1 = [tmp[i]]
            for j in range(i+1, len(s)):
                # 如果下一个字母不重复，则进行填充然后长度加一
                if tmp[j] not in tmp1:
                    tmp1.append(tmp[j])
                    max_len += 1
                    # 如果要添加的字母为最后一个位置的话，则记录当前长度
                    if j == len(s)-1:
                        max_list.append(max_len)
                # 如果下一个字母重复了
                else:
                    max_list.append(max_len)
                    break

        return max(max_list)
```

这回超时了。  
首先，一个很基础的概念：python 中字符串本身就是一个字符串列表，可以直接通过索引进行操作。不用非得转成list  
其次，每个窗口中的字符串都进行排序会占用很长的运行时长，因此字符串一旦很长就会超时  

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        p_list = []
        s_list = []
        for e in p:
            p_list.append(e)

        p_list.sort()

        for e in s:
            s_list.append(e)

        len_s = len(s_list)
        len_p = len(p_list)
        res = []
        for i in range(len_s-len_p+1):
            
            # 针对一个窗口中的字母内容逐个判断是否在 p 中
            # 注意不能只判断在不在，应该判断数量
            # 想法是 sample 和 p_list 分别排序，然后判断是否完全一致
            sample = []
            for j in range(len_p):
                sample.append(s_list[i+j])

            sample.sort()
            if sample == p_list:
                # 存放当前窗口起始索引
                res.append(i)

        return res
```

注意 counter 的使用方法

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:

        len_s = len(s)
        len_p = len(p)
        res = []
        window = Counter(s[:len_p])
        p_window = Counter(p)

        if window == p_window:
            res.append(0)

        for i in range(len_s-len_p):
            
            # 旧窗口删除第一个，添加即将进入窗口的元素
            window[s[i]] -= 1
            window[s[i+len_p]] += 1

            # 确保窗口中元素没有数量为 0 的，防止误判
            if window[s[i]] == 0:
                del window[s[i]]
            
            if window == p_window:
                # 存放当前窗口起始索引
                res.append(i+1)
            

        return res
```
