### 异常处理
1. 不做处理：整个程序因为一个Bug，中断执行
2. 捕获异常：按照我们自己的处理方式对异常处理，程序继续执行（做预案）
  
```Exception```是所有异常类型的父类，可以用它来捕获所有类型错误

````python
try:
    print("----------------")
    print(my_name)
    print("----------------")

except Exception as e:
    print("变量不存在，错误信息：", e)

finally:
    print("释放资源")
````

````python
try:
    print("----------------")
    print(my_name)
    print("----------------")

except NameError as e:
    print("变量不存在，错误信息：", e)

finally:
    print("释放资源")
````

````python
try:
    print("----------------")
    print(1 / 0)
    print("----------------")
    
except ZeroDivisionError as e:
    print("0不能做被除数，错误信息：", e)
finally:
    print("释放资源")
````

````python
try:
    print("----------------")
    print("ABC"[10])
    print("----------------")
    
except IndexError as e:
    print("超出索引，错误信息：", e)
finally:
    print("释放资源")
````
