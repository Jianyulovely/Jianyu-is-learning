### 装饰器的使用
装饰器会将被装饰的函数当作参数传递给装饰器同名的函数
1. 存在函数的嵌套关系
2. 内层函数引用了外层函数的变量
3. 外层函数返回内层函数的地址值

````python
def ready_send(send_func):

    def add_log():
        print("准备发送...")
        send_func()
        print("发送成功~")
    
    return add_log

@ready_send
def send_wechat():
    print("发微信！")

@ready_send
def send_msg():
    print("发短信！")
    
@ready_send
def send_tiktok():
    print("发抖音！")
````

````python
send_wechat()
send_msg()
send_tiktok()
````

### 装饰器的本质
编译器读取到```@ready_send```时候，先执行的就是ready_send这个函数，但是传入的参数是被装饰的函数```def send_wechat()```  
所以相当于执行的就是```ready_send(send_wechat)```  装饰器返回的函数```add_log```就是经过包装之后的```send_wechat```  

- 如果被装饰的函数有参数传入，那么需要在装饰器中对x的参数输入部分进行改动  
- 如果被装饰的函数有返回值的话，需要在装饰器内部对于自定义的函数添加返回内容。

````python
def ready_send(send_func):

    def add_log(*args, **kargs):
        print("准备发送...")
        send_func(*args, **kargs)
        print("发送成功~")
    
    return add_log

@ready_send
def send_wechat(user: str):
    print("发微信！", user)
    return 100

@ready_send
def send_msg(user: str, number: str):
    print("发短信！", user, number)
    
@ready_send
def send_tiktok(user: str, number: str, music: str):
    print("发抖音！", user, number, music)
````

````python
send_wechat("张三")
send_msg(user="李四", number="1388888")
send_tiktok(user="李四", number="1388888", music="Snooze")
````

这里返回值是None，证明原本的```send_wechat```被包装成了```add_log```

````python
print(send_wechat("Jason"))
````

在内部函数返回被装饰函数本身的返回值

````python
def ready_send_new(send_func):

    def add_log_value(*args, **kargs):
        print("准备发送...")
        send_val = send_func(*args, **kargs)
        print("发送成功~")

        return send_val
    
    return add_log_value

@ready_send_new
def send_wechat(user: str):
    print("发微信！", user)
    return 100

print(send_wechat("Gallary"))
````
