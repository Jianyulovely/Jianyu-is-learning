### 网络基础知识
- IP地址：设备在互联网中的地址（身份证），每个连入网络的设备都有一个IP地址。
    - IPV4：32位二进制  
    
    - IPV6：128位二进制  
    - **127.0.0.1** 表示本机地址（本地回环地址）

- 域名(Domain Name)，是由一串用点分隔的英文字母组成的。用于简化记忆IP地址(www.baidu.com)  
通过DNS（域名解析服务器）将域名和IP地址相互映射，便于记忆和访问。  
    - localhost为本机域名  
    

- 端口号(Port)：取值范围0-65535的整数，用来标识计算机设备中的运行中的程序。  
每个程序启动后都会占用一个端口  
比方说在同一个地址的设备中运行多个程序（都是www.baidu.com旗下的），只用IP地址访问对应的设备无法确定要哪个  
应用上的数据。  
    - 注意：HTTP协议默认端口号为**80**，HTTPS协议默认端口为**443**  
    所以，如果域名 **www.baidu.com** 对应的地址为 110.242.69.21，则在浏览器访问 **https://www.baidu.com/** 的时候其实  
    访问的是**110.242.69.21:443**

### 网络模型
为了便于对网络中传输的数据进行管理，国际ISO组织同意了程序在网路中通信的模型和标准。包括
- OSI网络模型：全球网络互联标准模型
- TCP/IP网络模型：认为是OSI的简化版  


### HTTP协议
- Hyper Text Transfer Protocol, 超文本传输协议，规定了客户端和服务器之间数据传输的规则  
只有在请求及响应中都遵循了统一的规则，服务端才能读懂客户端发来的请求，客户端才能解析服务端响应的结果。

- 特点：
    1. 基于文本的协议：请求和响应的部分的协议内容为文本格式，底层通过TCP协议传输，稳定性强
    2. 基于请求响应模型：一次请求对应一次响应，必须由客户端先发起请求，服务端才会返回响应
    3. 无状态：服务端不会记忆与客户端的历史交互信息，每次请求都是独立的

#### HTTP-请求数据格式
请求行：请求方式、资源路径、协议
- 请求方式
    - GET: 请求参数在请求行中，没有请求体。如：```/api/corses?name=python&status=1```。GET请求参数大小在浏览器中是有限制的。  
    参数之间使用```&```符号进行分隔
    - POST: 请求参数在请求体中，POST请求大小没有限制（图片、视频传输得用POST）  
    另外，由于对于大模型的请求的内容有很多文本（很长的提示词什么的），所以GET参数大小不足进行请求，只能用POST

请求头：格式 key-value  
请求体：请求参数部分  


### 响应数据格式
响应行：协议、状态码  
- 状态码：
    - 200：客户端请求成功
    - 400：客户端请求参数错误
    - 404：请求资源不存在，url输入有误，或者网站资源被删除 404 Not Found
    - 500：服务器发生了不可预期错误

响应头：key-value格式  
响应体：存放服务器响应的数据
